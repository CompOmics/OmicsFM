# ProtGPT — bulk transcriptomics SST UMAP

Whole-sample (SST) embeddings from the bulk-transcriptomics model (`bulk_trans_esmc`),
reduced with UMAP and coloured by tissue. Drawn from the **training corpus** (614k samples)
with a balanced per-tissue cap, over a 46-tissue map with the GI tract and brain split into
their sub-regions. Mirrors `api.ipynb` (proteomics) and `sc_transcript.ipynb` (single-cell).

```python
%load_ext autoreload
%autoreload 2

import os
# run from the repo root so the checkpoint's relative fasta/ and esmc_cache paths resolve
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")
print("cwd:", os.getcwd())

import numpy as np
import anndata as ad
from protgpt.api import compute_sst, visualize_sst

CKPT  = "model/bulk_trans_esmc/best_model.ckpt"
DATA  = "data/bulk_trans/protein_mapped/train.h5ad"   # full training corpus (614k samples)
FASTA = "fasta/human_proteome_canonical_31032022.fasta"
ESMC  = "data/esmc_human_cache.pt"
```

## 1. Tissue classes + balanced subsample
46 tissues, with the **GI tract** (esophagus / stomach / small intestine / colon / rectum) and
the **brain** (cortical lobes / cerebellum / substantia nigra / spinal cord / retina) split into
sub-regions. Fine tissues not in `TISSUE_MAP` are dropped. We then take up to `CAP` samples per
tissue from the 614k train split (set `CAP=None` for all mapped samples).

```python
TISSUE_MAP = {
    # blood / immune
    'blood':            ['blood', 'blood serum', 'blood plasma'],
    'PBMCs':            ['PBMCs'],
    'T cells':          ['T cells', 'CD4+ T cells', 'CD8+ T cells'],
    'B cells':          ['B cells'],
    'NK cells':         ['NK cells'],
    'monocytes':        ['monocytes', 'macrophage'],
    'neutrophil':       ['neutrophil'],
    'platelets':        ['platelets'],
    'bone marrow':      ['bone marrow'],
    'lymph node':       ['lymph node', 'tonsil'],
    'spleen':           ['spleen'],
    # nervous system (split)
    'brain':            ['brain'],
    'frontal cortex':   ['frontal cortex'],
    'temporal lobe':    ['temporal lobe'],
    'occipital lobe':   ['occipital lobe'],
    'parietal lobe':    ['parietal lobe'],
    'cerebellum':       ['cerebellum'],
    'substantia nigra': ['substantia nigra'],
    'spinal cord':      ['spinal cord'],
    'retina':           ['retina'],
    # cardiovascular / respiratory
    'heart':            ['heart'],
    'aorta':            ['aorta'],
    'lung':             ['lung'],
    # GI tract (split) + digestive organs
    'esophagus':        ['esophagus'],
    'stomach':          ['stomach'],
    'small intestine':  ['small intestine', 'duodenum'],
    'colon':            ['colon', 'cecum'],
    'rectum':           ['rectum'],
    'liver':            ['liver'],
    'pancreas':         ['pancreas'],
    # endocrine
    'thyroid':          ['thyroid gland'],
    'adrenal gland':    ['adrenal gland'],
    # urinary
    'kidney':           ['kidney', 'cortex of kidney'],
    'urinary bladder':  ['urinary bladder'],
    # reproductive
    'prostate':         ['prostate'],
    'testis':           ['testis', 'sperm', 'seminal plasma'],
    'ovary':            ['ovary'],
    'uterus':           ['uterine endometrium', 'uterus'],
    'cervix':           ['cervix'],
    'placenta':         ['placenta', 'trophoblast'],
    'breast':           ['breast'],
    # connective / skin / other
    'adipose tissue':   ['adipose tissue'],
    'skin':             ['skin'],
    'skeletal muscle':  ['skeletal muscle'],
    'bone':             ['bone'],
    'umbilical cord':   ['umbilical cord'],
}
fine_to_class = {fine: cls for cls, fines in TISSUE_MAP.items() for fine in fines}
print(f"{len(TISSUE_MAP)} tissue classes")
```

```python
# balanced subsample of the train corpus: up to CAP samples per tissue (set CAP=None for all)
CAP, SEED = 4000, 0

obs_all = ad.read_h5ad(DATA, backed="r").obs
tc_all = obs_all["tissue_harmonized"].astype(str).map(fine_to_class).to_numpy()
rng = np.random.default_rng(SEED)
sel = []
for cls in TISSUE_MAP:                                  # iterate in legend order
    pos = np.where(tc_all == cls)[0]
    if CAP is not None and pos.size > CAP:
        pos = rng.choice(pos, CAP, replace=False)
    sel.append(pos)
cap_idx = np.sort(np.concatenate(sel))
print(f"selected {len(cap_idx):,} / {len(obs_all):,} train samples "
      f"across {len(set(tc_all[cap_idx]))} tissues")
```

## 2. `compute_sst` on the subsample
ESM-C model -> pass the FASTA + ESM-C cache. SST is computed only on the capped subset
(embedding all 614k would be far too heavy); bins are built on the fly for this subset.

```python
adata = ad.read_h5ad(DATA, backed="r")[cap_idx].to_memory()
out = compute_sst(CKPT, adata, device="cuda", batch_size=64, num_workers=0,
                  fasta_path=FASTA, esmc_cache=ESMC)

# bulk obs carries a 'sample_id' column AND index -> drop the column to avoid an ambiguous merge
obs = adata.obs.copy().drop(columns="sample_id", errors="ignore")
obs["tissue_class"] = obs["tissue_harmonized"].astype(str).map(fine_to_class)
mapped = obs["tissue_class"].notna()
print("SST:", out["sst_emb"].shape, "| classes:", int(obs.loc[mapped, "tissue_class"].nunique()))
```

```python
keep = set(obs.index[mapped])
m = np.array([sid in keep for sid in out["sample_ids"]])
out_mapped = {"sst_emb": out["sst_emb"][m],
              "sample_ids": [s for s, k in zip(out["sample_ids"], m) if k]}
print(f"plotting {m.sum():,} of {len(m):,} samples across "
      f"{obs.loc[mapped, 'tissue_class'].nunique()} tissues")
```

```python
# 46 tissues — explicit colour per class, in legend order. TUNE freely. blood = red.
CLASS_COLORS = {
    "blood":             "#ed4242",
    "PBMCs":             "#bd311c",
    "T cells":           "#ed6f42",
    "B cells":           "#bd5b1c",
    "NK cells":          "#ed9b42",
    "monocytes":         "#bd851c",
    "neutrophil":        "#edc842",
    "platelets":         "#bdaf1c",
    "bone marrow":       "#e6ed42",
    "lymph node":        "#a1bd1c",
    "spleen":            "#b9ed42",
    "brain":             "#77bd1c",
    "frontal cortex":    "#8ded42",
    "temporal lobe":     "#4dbd1c",
    "occipital lobe":    "#60ed42",
    "parietal lobe":     "#23bd1c",
    "cerebellum":        "#42ed51",
    "substantia nigra":  "#1cbd3f",
    "spinal cord":       "#42ed7e",
    "retina":            "#1cbd69",
    "heart":             "#42edaa",
    "aorta":             "#1cbd93",
    "lung":              "#42edd7",
    "esophagus":         "#1cbdbd",
    "stomach":           "#42d7ed",
    "small intestine":   "#1c93bd",
    "colon":             "#42aaed",
    "rectum":            "#1c69bd",
    "liver":             "#427eed",
    "pancreas":          "#1c3fbd",
    "thyroid":           "#4251ed",
    "adrenal gland":     "#231cbd",
    "kidney":            "#6042ed",
    "urinary bladder":   "#4d1cbd",
    "prostate":          "#8d42ed",
    "testis":            "#771cbd",
    "ovary":             "#b942ed",
    "uterus":            "#a11cbd",
    "cervix":            "#e642ed",
    "placenta":          "#bd1caf",
    "breast":            "#ed42c8",
    "adipose tissue":    "#bd1c85",
    "skin":              "#ed429b",
    "skeletal muscle":   "#bd1c5b",
    "bone":              "#ed426f",
    "umbilical cord":    "#bd1c31",
}
order = [c for c in CLASS_COLORS if c in set(obs.loc[mapped, "tissue_class"])]
palette = [CLASS_COLORS[c] for c in order]

fig = visualize_sst(
    out_mapped,
    metadata=obs,                 # carries the new 'tissue_class' column
    color_by="tissue_class",
    color_order=order,
    palette=palette,
    hover=["tissue_harmonized", "disease_harmonized", "series_id"],
    method="umap",
    width=1600, height=1200,
    n_neighbors=100, min_dist=0.7, random_state=42, metric="euclidean",
)
```
